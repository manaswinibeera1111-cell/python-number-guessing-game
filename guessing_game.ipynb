```python
import random
number=random.randint(1,100)
guess=0
attempts=0
while guess!=number:
  guess=int(input("Enter your guess:"))
  attempts=attempts+1
  if(guess<number):
    print("Too low")
  elif(guess>number):
    print("Too high")
  else:
    print("You guessed the number")
    print("Attempts:",attempts)
```
